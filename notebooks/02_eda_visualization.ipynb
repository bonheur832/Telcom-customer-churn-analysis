```python
# Telco Customer Churn - EDA & Visualization

# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set visualization style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline

# Set figure size
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

# Load the cleaned data
df = pd.read_csv('../data/cleaned_customer_churn.csv')

print("="*60)
print("EDA & VISUALIZATION - DATASET OVERVIEW")
print("="*60)
print(f"Dataset Shape: {df.shape}")
print(f"Number of Records: {len(df)}")
print(f"Number of Features: {len(df.columns)}")
print("\nColumns:", df.columns.tolist())

# Target variable distribution
print("\n" + "="*60)
print("TARGET VARIABLE DISTRIBUTION")
print("="*60)

# Check churn columns
churn_cols = [col for col in df.columns if 'churn' in col.lower()]
print(f"Churn-related columns: {churn_cols}")

# Focus on Churn Value (binary target)
if 'Churn Value' in df.columns:
    churn_distribution = df['Churn Value'].value_counts()
    churn_percentage = df['Churn Value'].value_counts(normalize=True) * 100

    print("\nChurn Distribution:")
    print(churn_distribution)
    print("\nChurn Percentage:")
    print(churn_percentage)

    # Visualization
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # Count plot
    sns.countplot(x='Churn Value', data=df, ax=axes[0])
    axes[0].set_title('Churn Distribution - Count Plot')
    axes[0].set_xlabel('Churn (0=No, 1=Yes)')
    axes[0].set_ylabel('Count')

    # Add count labels
    for i, count in enumerate(churn_distribution.values):
        axes[0].text(i, count + 50, str(count), ha='center', va='bottom', fontsize=12)

    # Pie chart
    axes[1].pie(churn_distribution.values, labels=['No Churn', 'Churn'],
                autopct='%1.1f%%', startangle=90, colors=['#2ecc71', '#e74c3c'])
    axes[1].set_title('Churn Distribution - Pie Chart')

    plt.tight_layout()
    plt.savefig('../reports_2/churn_distribution.png', dpi=300, bbox_inches='tight')
    plt.show()

    print(f"\nOverall Churn Rate: {df['Churn Value'].mean()*100:.2f}%")

# Demographic features analysis
print("\n" + "="*60)
print("DEMOGRAPHIC ANALYSIS")
print("="*60)

# Identify demographic columns
demographic_cols = ['Gender', 'Senior Citizen', 'Partner', 'Dependents']
demographic_cols = [col for col in demographic_cols if col in df.columns]

if demographic_cols:
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    axes = axes.flatten()

    for idx, col in enumerate(demographic_cols):
        if idx < len(axes):
            # Create count plot with churn overlay
            sns.countplot(x=col, hue='Churn Value', data=df, ax=axes[idx])
            axes[idx].set_title(f'{col} Distribution by Churn')
            axes[idx].set_xlabel(col)
            axes[idx].set_ylabel('Count')
            axes[idx].legend(title='Churn', labels=['No', 'Yes'])

            # Calculate churn rates
            churn_rates = df.groupby(col)['Churn Value'].mean() * 100
            print(f"\n{col} Churn Rates:")
            print(churn_rates)

    plt.tight_layout()
    plt.savefig('../reports_2/demographic_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()

# Service usage analysis
print("\n" + "="*60)
print("SERVICE USAGE ANALYSIS")
print("="*60)

# Identify service columns
service_cols = ['Phone Service', 'Multiple Lines', 'Internet Service',
                'Online Security', 'Online Backup', 'Device Protection',
                'Tech Support', 'Streaming TV', 'Streaming Movies']
service_cols = [col for col in service_cols if col in df.columns]

# Calculate churn rates for each service
service_churn_rates = {}
for col in service_cols:
    if df[col].nunique() <= 5:  # Only analyze columns with few unique values
        churn_rate = df.groupby(col)['Churn Value'].mean().sort_values(ascending=False)
        service_churn_rates[col] = churn_rate
        print(f"\n{col} Churn Rates:")
        print(churn_rate)

# Visualize service churn rates
if service_churn_rates:
    n_cols = len(service_churn_rates)
    n_rows = (n_cols + 2) // 3

    fig, axes = plt.subplots(n_rows, 3, figsize=(18, 5*n_rows))
    axes = axes.flatten()

    for idx, (service, rates) in enumerate(service_churn_rates.items()):
        if idx < len(axes):
            bars = axes[idx].bar(range(len(rates)), rates.values * 100, color='#3498db')
            axes[idx].set_title(f'{service} Churn Rates')
            axes[idx].set_xlabel(service)
            axes[idx].set_ylabel('Churn Rate (%)')
            axes[idx].set_xticks(range(len(rates)))
            axes[idx].set_xticklabels(rates.index, rotation=45, ha='right')

            # Add value labels on bars
            for bar, rate in zip(bars, rates.values * 100):
                height = bar.get_height()
                axes[idx].text(bar.get_x() + bar.get_width()/2., height + 0.5,
                              f'{rate:.1f}%', ha='center', va='bottom', fontsize=10)

    # Hide empty subplots
    for idx in range(len(service_churn_rates), len(axes)):
        axes[idx].axis('off')

    plt.tight_layout()
    plt.savefig('../reports_2/service_churn_rates.png', dpi=300, bbox_inches='tight')
    plt.show()

# Contract and billing analysis
print("\n" + "="*60)
print("CONTRACT & BILLING ANALYSIS")
print("="*60)

contract_billing_cols = ['Contract', 'Paperless Billing', 'Payment Method']
contract_billing_cols = [col for col in contract_billing_cols if col in df.columns]

if contract_billing_cols:
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))

    for idx, col in enumerate(contract_billing_cols):
        if idx < 3:
            # Calculate churn rates
            churn_rates = df.groupby(col)['Churn Value'].mean().sort_values(ascending=False)

            # Create horizontal bar plot
            bars = axes[idx].barh(range(len(churn_rates)), churn_rates.values * 100,
                                 color=['#e74c3c' if i == 0 else '#3498db' for i in range(len(churn_rates))])
            axes[idx].set_title(f'{col} Churn Rates')
            axes[idx].set_xlabel('Churn Rate (%)')
            axes[idx].set_yticks(range(len(churn_rates)))
            axes[idx].set_yticklabels(churn_rates.index)

            # Add value labels
            for i, (bar, rate) in enumerate(zip(bars, churn_rates.values * 100)):
                axes[idx].text(rate + 0.5, bar.get_y() + bar.get_height()/2,
                              f'{rate:.1f}%', ha='left', va='center', fontsize=10)

            print(f"\n{col} Churn Rates:")
            print(churn_rates * 100)

    plt.tight_layout()
    plt.savefig('../reports_2/contract_billing_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()

# Numerical features analysis
print("\n" + "="*60)
print("NUMERICAL FEATURES ANALYSIS")
print("="*60)

# Identify numerical columns
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
# Remove target variable if present
if 'Churn Value' in numerical_cols:
    numerical_cols.remove('Churn Value')
if 'CustomerID' in numerical_cols:
    numerical_cols.remove('CustomerID')

print(f"Numerical columns to analyze: {numerical_cols}")

if numerical_cols:
    # Create distribution plots
    n_cols = len(numerical_cols)
    n_rows = (n_cols + 2) // 3

    fig, axes = plt.subplots(n_rows, 3, figsize=(18, 5*n_rows))
    axes = axes.flatten()

    for idx, col in enumerate(numerical_cols):
        if idx < len(axes):
            # Create subplot with distribution and boxplot
            ax1 = axes[idx]

            # Histogram with KDE
            sns.histplot(data=df, x=col, hue='Churn Value', kde=True,
                        element='step', stat='density', common_norm=False, ax=ax1)
            ax1.set_title(f'{col} Distribution by Churn')
            ax1.set_xlabel(col)
            ax1.set_ylabel('Density')
            ax1.legend(title='Churn', labels=['No', 'Yes'])

            # Calculate statistics
            churn_stats = df.groupby('Churn Value')[col].describe()
            print(f"\n{col} Statistics by Churn:")
            print(churn_stats)

    # Hide empty subplots
    for idx in range(len(numerical_cols), len(axes)):
        axes[idx].axis('off')

    plt.tight_layout()
    plt.savefig('../reports_2/numerical_features_distribution.png', dpi=300, bbox_inches='tight')
    plt.show()

# Correlation analysis
print("\n" + "="*60)
print("CORRELATION ANALYSIS")
print("="*60)

# Select numerical columns for correlation
correlation_cols = numerical_cols + ['Churn Value'] if 'Churn Value' in df.columns else numerical_cols

if len(correlation_cols) > 1:
    # Calculate correlation matrix
    corr_matrix = df[correlation_cols].corr()

    # Plot correlation heatmap
    fig, ax = plt.subplots(figsize=(12, 10))

    # Create mask for upper triangle
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

    # Plot heatmap
    sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
                center=0, square=True, linewidths=0.5, cbar_kws={"shrink": 0.8}, ax=ax)

    ax.set_title('Correlation Matrix of Numerical Features')
    plt.tight_layout()
    plt.savefig('../reports_2/correlation_matrix.png', dpi=300, bbox_inches='tight')
    plt.show()

    # Get correlations with target
    if 'Churn Value' in corr_matrix.columns:
        target_corr = corr_matrix['Churn Value'].sort_values(ascending=False)
        print("\nCorrelation with Churn Value:")
        print(target_corr)

        # Plot correlations with target
        fig, ax = plt.subplots(figsize=(10, 8))
        target_corr_no_self = target_corr.drop('Churn Value', errors='ignore')
        colors = ['red' if x > 0 else 'blue' for x in target_corr_no_self.values]
        bars = ax.barh(range(len(target_corr_no_self)), target_corr_no_self.values, color=colors)
        ax.set_yticks(range(len(target_corr_no_self)))
        ax.set_yticklabels(target_corr_no_self.index)
        ax.set_xlabel('Correlation Coefficient')
        ax.set_title('Feature Correlation with Churn Value')
        ax.axvline(x=0, color='black', linestyle='-', linewidth=0.5)

        # Add value labels
        for i, (bar, corr) in enumerate(zip(bars, target_corr_no_self.values)):
            ax.text(corr + (0.01 if corr >= 0 else -0.05), bar.get_y() + bar.get_height()/2,
                   f'{corr:.3f}', ha='left' if corr >= 0 else 'right', va='center', fontsize=10)

        plt.tight_layout()
        plt.savefig('../reports_2/target_correlation.png', dpi=300, bbox_inches='tight')
        plt.show()

# Tenure analysis
print("\n" + "="*60)
print("TENURE ANALYSIS")
print("="*60)

if 'Tenure Months' in df.columns:
    # Create tenure groups
    df['Tenure_Group'] = pd.cut(df['Tenure Months'],
                                bins=[0, 12, 24, 36, 48, 60, 72],
                                labels=['0-1yr', '1-2yr', '2-3yr', '3-4yr', '4-5yr', '5-6yr'])

    # Calculate churn rate by tenure group
    tenure_churn = df.groupby('Tenure_Group')['Churn Value'].mean() * 100

    print("\nChurn Rate by Tenure Group:")
    print(tenure_churn)

    # Visualization
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # Bar plot of churn rate by tenure group
    bars = axes[0].bar(range(len(tenure_churn)), tenure_churn.values, color='#9b59b6')
    axes[0].set_title('Churn Rate by Tenure Group')
    axes[0].set_xlabel('Tenure Group')
    axes[0].set_ylabel('Churn Rate (%)')
    axes[0].set_xticks(range(len(tenure_churn)))
    axes[0].set_xticklabels(tenure_churn.index, rotation=45)

    # Add value labels
    for bar, rate in zip(bars, tenure_churn.values):
        height = bar.get_height()
        axes[0].text(bar.get_x() + bar.get_width()/2., height + 0.5,
                    f'{rate:.1f}%', ha='center', va='bottom', fontsize=10)

    # Scatter plot: Tenure vs Monthly Charges colored by Churn
    scatter = axes[1].scatter(df['Tenure Months'], df['Monthly Charges'],
                             c=df['Churn Value'], cmap='coolwarm', alpha=0.6, s=50)
    axes[1].set_title('Tenure vs Monthly Charges (Colored by Churn)')
    axes[1].set_xlabel('Tenure (Months)')
    axes[1].set_ylabel('Monthly Charges ($)')
    axes[1].grid(True, alpha=0.3)

    # Add colorbar
    cbar = plt.colorbar(scatter, ax=axes[1])
    cbar.set_label('Churn (0=No, 1=Yes)')
    cbar.set_ticks([0, 1])
    cbar.set_ticklabels(['No', 'Yes'])

    plt.tight_layout()
    plt.savefig('../reports_2/tenure_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()

# Charges analysis
print("\n" + "="*60)
print("CHARGES ANALYSIS")
print("="*60)

charges_cols = ['Monthly Charges', 'Total Charges']
charges_cols = [col for col in charges_cols if col in df.columns]

if charges_cols:
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    for idx, col in enumerate(charges_cols):
        if idx < 2:
            # Create violin plot
            sns.violinplot(x='Churn Value', y=col, data=df, ax=axes[idx], palette='Set2')
            axes[idx].set_title(f'{col} Distribution by Churn')
            axes[idx].set_xlabel('Churn (0=No, 1=Yes)')
            axes[idx].set_ylabel(col)

            # Calculate statistics
            churn_0_stats = df[df['Churn Value'] == 0][col].describe()
            churn_1_stats = df[df['Churn Value'] == 1][col].describe()

            print(f"\n{col} Statistics:")
            print(f"Non-Churners: Mean=${churn_0_stats['mean']:.2f}, Median=${churn_0_stats['50%']:.2f}")
            print(f"Churners: Mean=${churn_1_stats['mean']:.2f}, Median=${churn_1_stats['50%']:.2f}")

    plt.tight_layout()
    plt.savefig('../reports_2/charges_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()

    # Create charge ratio analysis if both columns exist
    if 'Monthly Charges' in df.columns and 'Total Charges' in df.columns:
        # Calculate average monthly charge (Total Charges / Tenure Months)
        df['Avg_Monthly_Charge'] = df['Total Charges'] / df['Tenure Months'].replace(0, 1)

        # Compare actual vs average monthly charges
        fig, ax = plt.subplots(figsize=(12, 8))

        # Scatter plot
        scatter = ax.scatter(df['Monthly Charges'], df['Avg_Monthly_Charge'],
                            c=df['Churn Value'], cmap='RdYlBu_r', alpha=0.6, s=50)

        # Add diagonal line (where Monthly Charges = Avg Monthly Charge)
        max_val = max(df['Monthly Charges'].max(), df['Avg_Monthly_Charge'].max())
        ax.plot([0, max_val], [0, max_val], 'k--', alpha=0.5, label='y=x')

        ax.set_title('Monthly Charges vs Average Monthly Charge (Colored by Churn)')
        ax.set_xlabel('Monthly Charges ($)')
        ax.set_ylabel('Average Monthly Charge ($)')
        ax.grid(True, alpha=0.3)

        # Add colorbar
        cbar = plt.colorbar(scatter, ax=ax)
        cbar.set_label('Churn (0=No, 1=Yes)')
        cbar.set_ticks([0, 1])
        cbar.set_ticklabels(['No', 'Yes'])

        plt.legend()
        plt.tight_layout()
        plt.savefig('../reports_2/charges_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()

# Generate comprehensive EDA insights
print("\n" + "="*60)
print("EDA KEY INSIGHTS & SUMMARY")
print("="*60)

# Calculate key metrics
overall_churn_rate = df['Churn Value'].mean() * 100 if 'Churn Value' in df.columns else None

# Find top factors influencing churn
top_factors = []

# Check demographic factors
for col in ['Senior Citizen', 'Partner', 'Dependents']:
    if col in df.columns:
        churn_by_group = df.groupby(col)['Churn Value'].mean() * 100
        if len(churn_by_group) == 2:
            diff = abs(churn_by_group.iloc[0] - churn_by_group.iloc[1])
            top_factors.append((col, diff))

# Check service factors
for col in service_cols:
    if col in df.columns and df[col].nunique() <= 5:
        churn_by_service = df.groupby(col)['Churn Value'].mean() * 100
        if len(churn_by_service) > 1:
            diff = churn_by_service.max() - churn_by_service.min()
            top_factors.append((col, diff))

# Sort by impact
top_factors.sort(key=lambda x: x[1], reverse=True)

# Generate insights report
insights = f"""
COMPREHENSIVE EDA INSIGHTS REPORT
==================================

1. OVERALL CHURN RATE:
   - Overall churn rate: {overall_churn_rate:.2f}%
   - Non-churn customers: {(df['Churn Value'] == 0).sum() if 'Churn Value' in df.columns else 'N/A'}
   - Churn customers: {(df['Churn Value'] == 1).sum() if 'Churn Value' in df.columns else 'N/A'}

2. KEY DRIVERS OF CHURN (Top {min(5, len(top_factors))} factors):
"""

for i, (factor, impact) in enumerate(top_factors[:5]):
    insights += f"   {i+1}. {factor}: {impact:.1f}% difference in churn rates\n"

insights += f"""
3. DEMOGRAPHIC INSIGHTS:
"""

# Add demographic insights
for col in ['Gender', 'Senior Citizen', 'Partner', 'Dependents']:
    if col in df.columns:
        churn_rates = df.groupby(col)['Churn Value'].mean() * 100
        insights += f"   - {col}: "
        for value, rate in churn_rates.items():
            insights += f"{value}={rate:.1f}% "
        insights += "\n"

insights += f"""
4. SERVICE-RELATED INSIGHTS:
"""

# Add service insights
for col in ['Internet Service', 'Contract', 'Payment Method']:
    if col in df.columns:
        churn_rates = df.groupby(col)['Churn Value'].mean().sort_values(ascending=False) * 100
        insights += f"   - {col}: Highest churn for '{churn_rates.index[0]}' ({churn_rates.iloc[0]:.1f}%)\n"

insights += f"""
5. NUMERICAL FEATURES INSIGHTS:
"""

# Add numerical insights
if 'Tenure Months' in df.columns:
    tenure_churn_corr = df['Tenure Months'].corr(df['Churn Value']) if 'Churn Value' in df.columns else None
    insights += f"   - Tenure Months: Correlation with churn = {tenure_churn_corr:.3f}\n"

if 'Monthly Charges' in df.columns:
    monthly_churn_corr = df['Monthly Charges'].corr(df['Churn Value']) if 'Churn Value' in df.columns else None
    insights += f"   - Monthly Charges: Correlation with churn = {monthly_churn_corr:.3f}\n"

insights += f"""
6. BUSINESS RECOMMENDATIONS:
   - Focus retention efforts on customers with month-to-month contracts
   - Review pricing strategy for customers with high monthly charges
   - Implement special programs for senior citizens who show higher churn rates
   - Improve customer service for internet service users, particularly fiber optic
   - Consider loyalty programs for long-tenure customers

7. VISUALIZATIONS GENERATED:
   - churn_distribution.png
   - demographic_analysis.png
   - service_churn_rates.png
   - contract_billing_analysis.png
   - numerical_features_distribution.png
   - correlation_matrix.png
   - target_correlation.png
   - tenure_analysis.png
   - charges_analysis.png
   - charges_comparison.png
"""

print(insights)

# Save insights to file
with open('../reports_2/eda_comprehensive_insights.txt', 'w') as f:
    f.write(insights)

print("\nComprehensive EDA insights saved to '../reports_2/eda_comprehensive_insights.txt'")

# Final summary
print("\n" + "="*60)
print("EDA & VISUALIZATION COMPLETE")
print("="*60)

summary = """
SUMMARY OF EDA & VISUALIZATION:
1. Completed comprehensive analysis of demographic features
2. Analyzed service usage patterns and their impact on churn
3. Examined contract and billing factors
4. Created distribution plots for numerical features
5. Generated correlation analysis
6. Produced detailed tenure and charges analysis
7. Saved all visualizations to 'reports_2/' folder
8. Generated comprehensive insights report

NEXT STEPS:
1. Use insights for feature engineering
2. Proceed to predictive modeling
3. Validate findings with statistical tests
"""

print(summary)
```
